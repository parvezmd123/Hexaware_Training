```python
from pyspark.sql.functions import when, col
# 61. 

df = df.withColumn(
    "price_band",
    when(col("unit_price") >= 50000, "Premium")
    .when(col("unit_price") >= 10000, "Standard")
    .otherwise("Budget")
)

# 62. 
df.groupBy("price_band").count().show()

# 63. 
df = df.withColumn(
    "order_size",
    when(col("quantity") >= 4, "Large")
    .otherwise("Small")
)

# 64. 
df.groupBy("order_size").count().show()

# 65.
df = df.withColumn("total_amount", col("quantity") * col("unit_price"))
df = df.withColumn(
    "revenue_band",
    when(col("total_amount") >= 100000, "High")
    .otherwise("Low")
)

# 66. 
df.createOrReplaceTempView("sales")

# 67. 
spark.sql("SELECT * FROM sales").show()

# 68.
spark.sql("""
SELECT city, COUNT(*) AS total_orders
FROM sales
GROUP BY city
""").show()

# 69.
spark.sql("""
SELECT category, COUNT(*) AS total_orders
FROM sales
GROUP BY category
""").show()

# 70. 
spark.sql("""
SELECT city,
SUM(quantity * unit_price) AS revenue
FROM sales
GROUP BY city
""").show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, rank, dense_rank, col

# 71. Find revenue by category using SQL.
spark.sql("""
SELECT category,
SUM(quantity * unit_price) AS revenue
FROM sales
GROUP BY category
""").show()

# 72. Find top 5 orders by revenue using SQL.
spark.sql("""
SELECT *,
(quantity * unit_price) AS revenue
FROM sales
ORDER BY revenue DESC
LIMIT 5
""").show()

# 73. Find completed orders using SQL.
spark.sql("""
SELECT *
FROM sales
WHERE order_status='Completed'
""").show()

# 74. Find average price by category using SQL.
spark.sql("""
SELECT category,
AVG(unit_price) AS avg_price
FROM sales
GROUP BY category
""").show()

# 75. Find city with highest revenue using SQL.
spark.sql("""
SELECT city,
SUM(quantity * unit_price) AS revenue
FROM sales
GROUP BY city
ORDER BY revenue DESC
LIMIT 1
""").show()

# Window specification
windowSpec = Window.orderBy(col("total_amount").desc())

# 76. Create row_number by revenue.
df.withColumn("row_num", row_number().over(windowSpec)).show()

# 77. Create rank by revenue.
df.withColumn("rank", rank().over(windowSpec)).show()

# 78. Create dense_rank by revenue.
df.withColumn("dense_rank", dense_rank().over(windowSpec)).show()

# 79. Find top 5 revenue orders.
df.orderBy(col("total_amount").desc()).limit(5).show()

# 80. Find highest revenue order.
df.orderBy(col("total_amount").desc()).limit(1).show()
```

```python
from pyspark.sql.functions import col, row_number, sum, lag, lead, dense_rank# 81. Find highest revenue order by city.

cityWindow = Window.partitionBy("city").orderBy(col("total_amount").desc())
df.withColumn("rn", row_number().over(cityWindow)) \
  .filter(col("rn") == 1) \
  .show()

# 82. Find highest revenue order by category.
catWindow = Window.partitionBy("category").orderBy(col("total_amount").desc())
df.withColumn("rn", row_number().over(catWindow)) \
  .filter(col("rn") == 1) \
  .show()

# 83. Create running total of revenue.
runningWindow = Window.orderBy("order_id")
df.withColumn(
    "running_total",
    sum("total_amount").over(runningWindow)
).show()

# 84. Use lag() on revenue.
df.withColumn(
    "prev_revenue",
    lag("total_amount").over(windowSpec)
).show()

# 85. Use lead() on revenue.
df.withColumn(
    "next_revenue",
    lead("total_amount").over(windowSpec)
).show()

# 86. Compare current revenue with previous revenue.
df.withColumn(
    "prev_revenue",
    lag("total_amount").over(windowSpec)
).withColumn(
    "difference",
    col("total_amount") - col("prev_revenue")
).show()

# 87. Find revenue difference.
df.withColumn(
    "revenue_diff",
    col("total_amount") -
    lag("total_amount").over(windowSpec)
).show()

# 88. Find top 2 orders per city.
df.withColumn(
    "rn",
    row_number().over(cityWindow)
).filter(col("rn") <= 2).show()

# 89. Find top 3 orders per category.
df.withColumn(
    "rn",
    row_number().over(catWindow)
).filter(col("rn") <= 3).show()

# 90. Generate revenue ranking report.
df.withColumn(
    "rank",
    dense_rank().over(windowSpec)
).select(
    "order_id",
    "product",
    "total_amount",
    "rank"
).show()
```

```python
# 91. Read source data.
source_df = df

# 92. Create calculated columns.
source_df = source_df.withColumn(
    "total_amount",
    col("quantity") * col("unit_price")
)

# 93. Filter completed orders.
completed_df = source_df.filter(col("order_status") == "Completed")

# 94. Calculate revenue.
completed_df.select(
    sum("total_amount")
).show()

# 95. Generate city revenue report.
completed_df.groupBy("city") \
    .agg(sum("total_amount").alias("revenue")) \
    .show()

# 96. Generate category revenue report.
completed_df.groupBy("category") \
    .agg(sum("total_amount").alias("revenue")) \
    .show()

# 97. Save output as Parquet.
completed_df.write.mode("overwrite").parquet("sales_parquet")

# 98. Read Parquet.
parquet_df = spark.read.parquet("sales_parquet")

# 99. Create SQL view from Parquet.
parquet_df.createOrReplaceTempView("sales_parquet_view")

# 100. Generate final business report.
spark.sql("""
SELECT category,
SUM(total_amount) AS revenue,
COUNT(*) AS total_orders
FROM sales_parquet_view
GROUP BY category
ORDER BY revenue DESC
""").show()
```
