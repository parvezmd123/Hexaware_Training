```python
from pyspark.sql import SparkSession

spark=SparkSession.builder \
      .appName("Working with EDA") \
      .getOrCreate()
```

```python
from pyspark.sql.functions import col
appointments_data = [
(1001, "Hyderabad", "Cardiology", "Apollo", 1500, "Completed"),
(1002, "Bangalore", "Neurology", "Yashoda", 2200, "Completed"),
(1003, "Mumbai", "Dermatology", "Care", 900, "Pending"),
(1004, "Delhi", "Orthopedics", "Max", 2500, "Completed"),
(1005, "Chennai", "Pediatrics", "Apollo", 1200, "Cancelled"),
(1006, "Hyderabad", "Cardiology", "Care", 3000, "Completed"),
(1007, "Bangalore", "Dermatology", "Apollo", 1000, "Completed"),
(1008, "Mumbai", "Neurology", "Max", 2600, "Pending"),
(1009, "Delhi", "Cardiology", "Yashoda", 2800, "Completed"),
(1010, "Chennai", "Orthopedics", "Care", 2400, "Completed"),
(1011, "Hyderabad", "Pediatrics", "Apollo", 1100, "Completed"),
(1012, "Bangalore", "Cardiology", "Max", 3200, "Completed"),
(1013, "Mumbai", "Pediatrics", "Yashoda", 1300, "Cancelled"),
(1014, "Delhi", "Neurology", "Apollo", 2700, "Completed"),
(1015, "Chennai", "Dermatology", "Care", 950, "Pending")
]
columns = [
"appointment_id",
"city",
"department",
"hospital",
"consultation_fee",
"status"
]
df = spark.createDataFrame(appointments_data, columns)
display(df)
```

```python
from pyspark.sql.functions import sum, count, col

display(df.groupBy("city").count())
```

```python
display(df.groupBy("city").agg(sum("consultation_fee").alias("revenue")))
```

```python
display(df.groupBy("department").agg(sum("consultation_fee").alias("revenue")))
```

```python
display(df.groupBy("status").count())
```

```python
display(df.groupBy("hospital").agg(sum("consultation_fee").alias("revenue")))
```

```python
display(df.select("appointment_id", "consultation_fee"))
```

```python
display(df.orderBy("appointment_id").select("appointment_id", "consultation_fee"))
```

```python
display(df.groupBy("department").count())
```

```python
display(df.filter(col("status") == "Completed")
          .groupBy("city")
          .count())
```

```python
display(
    df.filter(col("status") != "Cancelled")
      .groupBy("city")
      .agg(sum("consultation_fee").alias("revenue"))
)
```

```python
display(
    df.groupBy("city")
      .agg(sum("consultation_fee").alias("revenue"))
      .orderBy(col("revenue").desc())
      .limit(3)
)
```

```python
display(
    df.groupBy("department")
      .agg(sum("consultation_fee").alias("revenue"))
      .orderBy(col("revenue").desc())
      .limit(3)
)
```

```python
display(
    df.groupBy("hospital")
      .agg(sum("consultation_fee").alias("revenue"))
      .orderBy(col("revenue"))
      .limit(1)
)
```

```python
display(
    df.groupBy("hospital")
      .agg(count("*").alias("appointments"))
)
```
