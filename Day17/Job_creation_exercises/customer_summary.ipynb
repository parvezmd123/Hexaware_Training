```python
# Exercise 3 - customer_summary notebook
# Step 5 - Read the customer_table saved by customer_loader
# NOTE: Update table name if you used a fully qualified name in customer_loader
df = spark.table("customer_table")
print(f"Total customers loaded: {df.count()}")
display(df)
```

```python
from pyspark.sql.functions import count

# Step 6 - Calculate number of customers per city
city_summary = df.groupBy("city") \
                 .agg(count("customer_id").alias("customer_count")) \
                 .orderBy("customer_count", ascending=False)

display(city_summary)
print("Customer summary by city completed")
```
