```python
products = [
    (1, "Laptop", 75000),
    (2, "Phone", 25000),
    (3, "Tablet", 30000),
    (4, "Monitor", 15000)
]

columns = ["product_id", "product_name", "price"]

product_df = spark.createDataFrame(products, columns)

# Display products
display(product_df)
```

```python
product_df.createOrReplaceTempView("products")

# SQL Query
avg_price = spark.sql("""
SELECT AVG(price) AS average_price
FROM products
""")

# Display result
display(avg_price)
```
