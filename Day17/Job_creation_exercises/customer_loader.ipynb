```python
# Exercise 3 - Customer Data
customer_data = [
    (1, "Alice",   "Mumbai",    25),
    (2, "Bob",     "Delhi",     30),
    (3, "Charlie", "Mumbai",    28),
    (4, "Diana",   "Bangalore", 35),
    (5, "Eve",     "Delhi",     22)
]
columns = ["customer_id", "name", "city", "age"]

# Step 2 - Create DataFrame
df = spark.createDataFrame(customer_data, columns)
display(df)
```

```python
# Step 3 - Save DataFrame as a managed Delta table
# NOTE: Update 'catalog_name.schema_name' to match your Unity Catalog catalog and schema
# Example: df.write.mode("overwrite").saveAsTable("main.default.customer_table")
df.write.mode("overwrite").saveAsTable("customer_table")
print("customer_table saved successfully")
```
