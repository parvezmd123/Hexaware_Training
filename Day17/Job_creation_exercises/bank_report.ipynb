```python
# Exercise 4 - Banking Report
# Step 1 - Create account data
account_data = [
    (1001, "Alice",   "Mumbai",    150000),
    (1002, "Bob",     "Delhi",     200000),
    (1003, "Charlie", "Mumbai",     80000),
    (1004, "Diana",   "Bangalore", 320000),
    (1005, "Eve",     "Delhi",      50000)
]
columns = ["account_id", "customer_name", "city", "balance"]

# Step 2 - Create DataFrame
df = spark.createDataFrame(account_data, columns)
```

```python
from pyspark.sql.functions import sum as spark_sum

# Step 3 - Calculate total balance by city
report = df.groupBy("city") \
           .agg(spark_sum("balance").alias("total_balance")) \
           .orderBy("total_balance", ascending=False)

# Display the report
display(report)
print("Banking Report Generated")
```
